#### Problema (5)
Asistente Escolar con IA". Como el control de flujo se basa estrictamente en condiciones (if, elif, else) y operadores, la estructura funciona como un "árbol de decisión". La IA analizará tus respuestas para adivinar


```python
# Simulador de Asistente Escolar Inteligente (Basado en Reglas)

print("=========================================================")
print("🤖 BIENVENIDO AL ASISTENTE ESCOLAR CON IA (Versión Beta) 🤖")
print("=========================================================")
print("Este sistema analizará tu situación académica usando lógica condicional.")
print("---------------------------------------------------------\n")

# --- MÓDULO 1: DIAGNÓSTICO DEL PROFESOR ---
print("--- SECCIÓN 1: Perfil del Profesor ---")
print("¿Cómo dicta las clases tu profesor?")
print("A) Explicaciones claras, usa diapositivas y es justo al calificar.")
print("B) Explicaciones complejas, rudo al calificar y deja tareas imposibles.")
print("C) No se le entiende nada, falta mucho o solo lee PDFs.")
perfil_profesor = input("Elige una opción (A, B o C): ")

# IA procesa el perfil del profesor
if perfil_profesor == "A" or perfil_profesor == "a":
    tipo_profesor = "El Barco Justo (Exigente pero pasable)"
    estrategia_profesor = "Estudia de los apuntes de clase y entrega todo a tiempo. Tienes altas probabilidades de aprobar."
elif perfil_profesor == "B" or perfil_profesor == "b":
    tipo_profesor = "El Cuchilla (Filtrador de alumnos)"
    estrategia_profesor = "Busca libros externos, estudia en grupo y prepárate para exámenes sorpresa bien difíciles."
elif perfil_profesor == "C" or perfil_profesor == "c":
    tipo_profesor = "El Fantasma (Ausente o desinteresado)"
    estrategia_profesor = "Te toca ser autodidacta. Consigue exámenes de semestres pasados porque el examen final será una lotería."
else:
    tipo_profesor = "Desconocido (Opción no válida)"
    estrategia_profesor = "No ingresaste una opción válida, ten cuidado con las instrucciones."

print(f"\n[IA Analizando] -> Tipo de Profesor Detectado: {tipo_profesor}")
print(f"[IA Consejo] -> {estrategia_profesor}\n")
print("---------------------------------------------------------\n")


# --- MÓDULO 2: ESTADO DE LA MATERIA ---
print("--- SECCIÓN 2: Rendimiento en la Materia ---")
calificacion_actual = float(input("¿Cuál es tu calificación actual en la materia (0.0 a 10.0)?: "))
asistencias_porcentaje = int(input("¿Qué porcentaje de asistencia tienes (0 a 100)?: "))

# IA calcula tu estado usando operadores lógicos (and, or) y relacionales (>=, <)
print("\n[IA Computando Estado Académico...]")

if calificacion_actual >= 7.0 and asistencias_porcentaje >= 80:
    print("🏆 ESTADO: ¡ZONA DE CONFORT! Estás aprobado. Sigue así y no te confíes.")

elif calificacion_actual >= 5.0 and calificacion_actual < 7.0 and asistencias_porcentaje >= 80:
    print("⚠️ ESTADO: EN RIESGO REPARABLE. Estás en la cuerda floja.")
    print("Consejo: Un último esfuerzo en el examen final o proyecto te dará el aprobado.")

elif calificacion_actual >= 7.0 and asistencias_porcentaje < 80:
    print("🚨 ESTADO: PELIGRO POR FALTAS. Tienes buena nota, pero podrías reprobar por asistencia.")
    print("Consejo: Habla con el profesor urgentemente para justificar tus faltas o hacer un trabajo extra.")

elif calificacion_actual < 5.0 or asistencias_porcentaje < 60:
    print("💀 ESTADO: ALERTA ROJA / CRÍTICO.")

    # Evaluación anidada para dar una última alternativa extrema
    print("¿Hay opción de examen extraordinario o de recuperación al final del semestre?")
    tiene_extra = input("Responde (Sí/No): ")

    if tiene_extra == "Sí" or tiene_extra == "si" or tiene_extra == "SI" or tiene_extra == "s":
        print("-> Plan de Emergencia: Deja de lamentarte y empieza a estudiar YA para el examen extraordinario. Aún hay fe.")
    else:
        print("-> Diagnóstico Final: Matemáticamente estás fuera. Ve planeando recursar la materia el próximo semestre y elige mejor al profesor.")

else:
    print("Error: Los datos introducidos no tienen coherencia lógica.")

print("\n---------------------------------------------------------")
print("🤖 Fin de la simulación. ¡Buena suerte en tu semestre! 🤖")
print("=========================================================")
```
